# Calling the deployed Gemma inference app

This notebook is a client for the deployed Modal app; it does not define or deploy the model. First deploy the app from a terminal in this project with `modal deploy llm_inference_check_with_modal.py`, and authenticate the Modal CLI/SDK for the same environment. Run the model-download cell only when you need to populate or update the model Volume. Inference calls use the deployed class and pass each conversation as a list of chat messages.

```python
import modal

APP_NAME = "llm-t4-inference-experimentation"

download_model = modal.Function.from_name(APP_NAME, "download_model")
GemmaInference = modal.Cls.from_name(APP_NAME, "GemmaInference")
```

## Optional: populate or update the model Volume

Run this cell once when setting up the Volume or when you want to refresh the downloaded model. It invokes the deployed downloader, which needs the Hugging Face Secret configured on the Modal app.

```python
download_model.remote()
```

## Run inference

Provide a batch of conversations. Each conversation is a list of role/content messages. The deployed `run` method returns one decoded response per conversation.

```python
conversations: list[list[dict[str, str]]] = [
    [{"role": "user", "content": "What is 2 + 2?"}],
    [{"role": "user", "content": "Explain what a Python dictionary is in one sentence."}],
]

responses = GemmaInference().run.remote(conversations)

for index, response in enumerate(responses, start=1):
    print(f"Response {index}: {response}\n")
```
